```python
import pandas as pd
import glob
import os
```

```python
# Grab all 2024 taxi CSV files in the current directory
csv_files = glob.glob("yellow_tripdata_2024-*.csv")

# Combine them into one DataFrame
df_all = pd.concat((pd.read_csv(file, low_memory=False) for file in csv_files), ignore_index=True)

# Optional: Save the full combined dataset
df_all.to_csv("yellow_tripdata_2024_full_combined.csv", index=False)

# Preview shape
print(f"✅ Combined {len(csv_files)} files — Total rows: {df_all.shape[0]}")
df_all.head()
```

```python
# Convert to datetime
df_all['tpep_pickup_datetime'] = pd.to_datetime(df_all['tpep_pickup_datetime'], errors='coerce')

# Drop rows with invalid dates (if any)
df_all = df_all.dropna(subset=['tpep_pickup_datetime'])

# Extract time features
df_all['date'] = df_all['tpep_pickup_datetime'].dt.date
df_all['hour'] = df_all['tpep_pickup_datetime'].dt.hour
df_all['month'] = df_all['tpep_pickup_datetime'].dt.month
```

```python
daily_trips = df_all.groupby('date').size().reset_index(name='trip_count')
daily_trips.to_csv("summary_daily_trips.csv", index=False)
```

```python
hourly_trips = df_all.groupby('hour').size().reset_index(name='trip_count')
hourly_trips.to_csv("summary_hourly_trips.csv", index=False)
```

```python
hourly_fares = df_all.groupby('hour')['fare_amount'].mean().reset_index(name='avg_fare')
hourly_fares.to_csv("summary_hourly_avg_fare.csv", index=False)
```

```python
import pandas as pd

zone_lookup = pd.read_csv("taxi_zone_lookup.csv")
zone_map = zone_lookup.set_index("LocationID")["Borough"].to_dict()

# Define input/output file paths
input_file = "yellow_tripdata_2024_full_combined.csv"
output_file = "yellow_tripdata_2024_with_borough.csv"

# Prepare to write in chunks
chunksize = 500_000  # You can increase if your RAM allows

first_chunk = True
for chunk in pd.read_csv(input_file, chunksize=chunksize):
    # Map borough for pickup locations
    chunk['pickup_borough'] = chunk['PULocationID'].map(zone_map)

    # Save to new file (append mode)
    chunk.to_csv(output_file, mode='a', index=False, header=first_chunk)
    first_chunk = False  # Only write header for first chunk

    print(f"✅ Processed {len(chunk)} rows")
```

```python
# In a separate step — create a summary from the full combined file
zone_lookup = pd.read_csv("taxi_zone_lookup.csv")
zone_map = zone_lookup.set_index("LocationID")["Borough"].to_dict()

# Read only needed columns to reduce memory load
df = pd.read_csv("yellow_tripdata_2024_full_combined.csv", usecols=['DOLocationID'])

# Map drop-off borough
df['dropoff_borough'] = df['DOLocationID'].map(zone_map)

# Summarize trips by drop-off borough
dropoff_summary = df['dropoff_borough'].value_counts().reset_index()
dropoff_summary.columns = ['dropoff_borough', 'trip_count']
dropoff_summary.to_csv("summary_dropoffs_by_borough.csv", index=False)
```

```python
mta_df = pd.read_csv("MTA_Daily_Ridership_Data__2020_-_2025_20250411.csv")
mta_df['Date'] = pd.to_datetime(mta_df['Date'])

# Filter for 2024 only
mta_2024 = mta_df[mta_df['Date'].dt.year == 2024]

# Select subway ridership column
mta_2024 = mta_2024[['Date', 'Subways: Total Estimated Ridership']]
mta_2024.columns = ['date', 'subway_ridership']

# Save cleaned version
mta_2024.to_csv("summary_mta_subway_2024.csv", index=False)
```

```python
crash_df = pd.read_csv("TLC_Vehicles_Involved_in_Crashes__Local_Law_31__20250411.csv")

# Filter for Yellow Taxi and 2024
crash_yt = crash_df[(crash_df['industry'] == 'Yellow Taxi') & (crash_df['year_month'].str.startswith("2024"))]

# Clean and save
crash_summary = crash_yt[['year_month', 'total_crashes']]
crash_summary.to_csv("summary_crashes_yellow_taxi_2024.csv", index=False)
```

```python
traffic_df = pd.read_csv("Traffic_Volume_Counts_20250311.csv")

# Example: Average traffic volume by hour
hourly_columns = [col for col in traffic_df.columns if 'AM' in col or 'PM' in col]
traffic_df['average_volume'] = traffic_df[hourly_columns].mean(axis=1)

traffic_summary = traffic_df[['Roadway Name', 'From', 'To', 'Direction', 'average_volume']]
traffic_summary.to_csv("summary_traffic_volume_by_segment.csv", index=False)
```

```python
import pandas as pd

# File with pickup_borough column already included
input_file = "yellow_tripdata_2024_with_borough.csv"

# Initialize empty dictionary to accumulate counts
borough_counts = {}

# Read in chunks
chunksize = 500_000
for chunk in pd.read_csv(input_file, usecols=['pickup_borough'], chunksize=chunksize):
    counts = chunk['pickup_borough'].value_counts()
    for borough, count in counts.items():
        borough_counts[borough] = borough_counts.get(borough, 0) + count

# Convert to DataFrame
pickup_summary = pd.DataFrame(list(borough_counts.items()), columns=['pickup_borough', 'trip_count'])

# Save to CSV
pickup_summary.to_csv("summary_pickups_by_borough.csv", index=False)

print("✅ summary_pickups_by_borough.csv created successfully.")
```
