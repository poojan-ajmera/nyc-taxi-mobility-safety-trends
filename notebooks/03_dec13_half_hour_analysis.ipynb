```python
import pandas as pd

# Load December 2024 trip data
df = pd.read_csv("yellow_tripdata_2024-12.csv", parse_dates=['tpep_pickup_datetime'])
```

```python
# Filter just for December 13
df_dec13 = df[df['tpep_pickup_datetime'].dt.date == pd.to_datetime("2024-12-13").date()]
```

```python
df_dec13.loc[:, "hour"] = df_dec13["tpep_pickup_datetime"].dt.hour
```

```python
# Group by hour and calculate average fare, distance, congestion fee, and trip count
summary_by_hour = df_dec13.groupby('hour').agg({
    'trip_distance': 'mean',
    'fare_amount': 'mean',
    'congestion_surcharge': 'mean',
    'tpep_pickup_datetime': 'count'  # number of trips
}).rename(columns={'tpep_pickup_datetime': 'trip_count'}).reset_index()
```

```python
# Save summary
summary_by_hour.to_csv("summary_dec13_by_hour.csv", index=False)
```

```python
# Create 30-minute block column using .loc
df_dec13.loc[:, "time_block"] = df_dec13["tpep_pickup_datetime"].dt.floor("30min")

# Group by 30-minute block
summary_by_half_hour = df_dec13.groupby("time_block").agg({
    "trip_distance": "mean",
    "fare_amount": "mean",
    "congestion_surcharge": "mean",
    "tpep_pickup_datetime": "count"
}).rename(columns={"tpep_pickup_datetime": "trip_count"}).reset_index()

# Save summary
summary_by_half_hour.to_csv("summary_dec13_by_half_hour.csv", index=False)
```
